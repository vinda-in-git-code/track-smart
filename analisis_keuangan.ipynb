## DATA CLEANING

```python
data1.head()
```

```python
# mengganti format tanggal pada kolom `Date` menjadi format datetime dan menghapus timestamp
def preprocess_date(date_str):
    if pd.isna(date_str):
        return date_str
    
    date_str = str(date_str).strip()
    
    if " " in date_str:
        date_str = date_str.split(" ")[0]
        
    return date_str

data1['Date'] = data1['Date'].apply(preprocess_date)
data1['Date'] = pd.to_datetime(data1['Date'], dayfirst=True, format='mixed', errors='coerce')

# normalisasi untuk menghilangkan jam tetapi format tetap datetime
data1['Date'] = data1['Date'].dt.normalize()

data1['Date'].head()
```

```python
# melihat apakah ada missing values setelah tipe data pada kolom `Date` sudah diubah menjadi datetime
data1['Date'].isna().sum()
```

```python
# menghapus kolom yang tidak diperlukan untuk analisis
data1.drop(columns=['Mode', 'Subcategory', 'Note', 'Currency'], inplace=True)
data1.head()
```

```python
# mengganti nama kolom `Income/Expense` menjadi `Transaction Type`
data1.rename(columns={'Income/Expense': 'Transaction Type'}, inplace=True)
data1.head()
```

```python
print(data1.columns.tolist())
print(data1.head(3))
```

karena pada mata uang yang digunakan pada data1 masih dalam bentuk INR (Indian Rupee) sedangkan target pengguna website Finance Tracker adalah generasi muda Indonesia, maka nilai mata uangnya akan dikonversi ke dalam Rupiah mengikuti kurs pada tahun di dataset yaitu 2015-2018 menggunakan sumber kurs dari World Bank.

```python
# === KURS RATA-RATA TAHUNAN HISTORIS ===
# Sumber: World Bank, Official Exchange Rate (PA.NUS.FCRF)
# Indikator: PA.NUS.FCRF | https://data.worldbank.org
exchange_rates = {
    # tahun: (IDR_per_USD, INR_per_USD)
    2015: (13389, 64.15),
    2016: (13308, 67.20),
    2017: (13381, 65.12),
    2018: (14237, 68.40),
    2019: (14015, 70.42),
}

# ekstrak tahun dari kolom Date
data1['Year'] = pd.to_datetime(data1['Date']).dt.year

# Verifikasi tahun unik — pastikan sesuai dengan kurs yang tersedia
print("Tahun unik di data1:", data1['Year'].unique())

# fungsi konversi INR → IDR
def convert_inr_to_idr(amount, year):
    """
    Mengkonversi INR ke IDR menggunakan cross rate historis tahunan.
    Formula: IDR = INR × (IDR_per_USD ÷ INR_per_USD)
    Sumber: World Bank, PA.NUS.FCRF
    """
    if year not in exchange_rates:
        raise ValueError(f"Kurs untuk tahun {year} tidak tersedia.")
    
    idr_per_usd, inr_per_usd = exchange_rates[year]
    idr_per_inr = idr_per_usd / inr_per_usd
    return amount * idr_per_inr

# terapkan konversi
data1['Amount_IDR'] = data1.apply(
    lambda row: convert_inr_to_idr(row['Amount'], row['Year']),
    axis=1
)

data1.head()
```

nilai transaksi pada data1 telah dikonversi ke dalam bentuk rupiah dah disimpan pada kolom baru bernama `Amount_IDR`

```python
# menghapus kolom `Amount` dan `Year` pada data1 karena sudah tidak diperlukan lagi setelah konversi ke IDR
data1.drop(columns=['Amount', 'Year'], inplace=True)
data1.head()
```

```python
# mengecek kembali apakah ada kolom yang memiliki missing values dan duplikat setelah proses pembersihan dan konversi
print("Missing values per kolom:", data1.isna().sum())
print("Jumlah duplikat:", data1.duplicated().sum())
```

```python
# menghapus kolom yang duplikat
data1.drop_duplicates(inplace=True)
print("Jumlah Duplikat:", data1.duplicated().sum())
data1
```

### CLEANING DATA2

```python
data2.head()
```

```python
# menghapus kolom yang tidak diperlukan untuk proses analisis
data2.drop(columns = ['Description', 'Account Name', 'Month'], inplace=True)
data2.head()
```

```python
# mengganti isi kolom Transaction Type menjadi Expense/Income
data2['Transaction Type'] = data2['Transaction Type'].replace({'debit': 'Expense', 'credit': 'Income'})
data2.head()
```

```python
# melakukan konversi nilai mata uang pada data2 dari USD ke IDR menggunakan kurs rata-rata tahunan historis
exchange_rates = {
    # tahun: (IDR_per_USD, INR_per_USD)
    2015: (13389, 64.15),
    2016: (13308, 67.20),
    2017: (13381, 65.12),
    2018: (14237, 68.40),
    2019: (14015, 70.42),
}

# ekstrak tahun dari kolom Date
data2['Year'] = pd.to_datetime(data2['Date']).dt.year

# Verifikasi tahun unik — pastikan sesuai dengan kurs yang tersedia
print("Tahun unik di data2:", data2['Year'].unique())

def convert_usd_to_idr(amount, year):
    """
    Mengkonversi USD ke IDR menggunakan kurs historis tahunan.
    Formula: IDR = USD × IDR_per_USD
    Sumber: World Bank, PA.NUS.FCRF
    """
    if year not in exchange_rates:
        raise ValueError(f"Kurs untuk tahun {year} tidak tersedia.")
    
    idr_per_usd, _ = exchange_rates[year]
    return amount * idr_per_usd
        
# terapkan konversi
data2['Amount_IDR'] = data2.apply(
    lambda row: convert_usd_to_idr(row['Amount'], row['Year']),
    axis=1
)

data2.head()
```

```python
# menghapus kolom `Amount` dan `Year` yang sudah tidak diperlukan
data2.drop(columns=['Amount', 'Year'], inplace=True)
data2.head()
```

```python
# mengecek kembali missing values dan duplikat setelah proses pembersihan dan konversi
print("Missing values per kolom:", data2.isna().sum())
print("Jumlah duplikat:", data2.duplicated().sum())
```

setelah dilakukan pembersihan dan konversi, tidak terdapat missing values ataupun duplikasi pada data2. karena kedua data sudah dalam keadaan bersih, maka data siap untuk dilakukan merge.

```python
# melihat struktur data sebelum dilakukan merge
print("Kolom data1:", data1.columns.tolist())
print("Kolom data2", data2.columns.tolist())
print("Jumlah baris data1:", len(data1))
print("Jumlah baris data2:", len(data2))
```
